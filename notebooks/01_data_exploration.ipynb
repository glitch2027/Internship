```
import pandas as pd
import glob
import os
```

```
# Load all CSV datasets from data/raw
raw_dir = '../data/raw'
csv_files = glob.glob(os.path.join(raw_dir, '*.csv'))

datasets = {}
for file in csv_files:
    name = os.path.basename(file)
    datasets[name] = pd.read_csv(file)
```

```
# Print .shape, .dtypes, and .head() for each
for name, df in datasets.items():
    print(f'--- {name} ---')
    print(f'Shape: {df.shape}')
    print(f'Data Types:\n{df.dtypes}')
    print(f'Head:\n{df.head()}\n')
```

```
# Explore fund master
if 'fund_master.csv' in datasets:
    df = datasets['fund_master.csv']
    print('Unique Fund Houses:', df['fund_house'].unique() if 'fund_house' in df.columns else 'N/A')
    print('Unique Categories:', df['category'].unique() if 'category' in df.columns else 'N/A')
    print('Unique Sub-Categories:', df['sub_category'].unique() if 'sub_category' in df.columns else 'N/A')
    print('Unique Risk Grades:', df['risk_grade'].unique() if 'risk_grade' in df.columns else 'N/A')
```

```
# Validate AMFI codes
if 'fund_master.csv' in datasets and 'nav_history.csv' in datasets:
    master_codes = set(datasets['fund_master.csv']['amfi_code'].dropna().astype(str))
    nav_codes = set(datasets['nav_history.csv']['amfi_code'].dropna().astype(str))
    missing = master_codes - nav_codes
    print(f'Codes in fund_master missing in nav_history: {len(missing)}')
    if len(missing) == 0:
        print('All AMFI codes validate successfully!')
    else:
        print(f'Data Quality Issue: {len(missing)} codes missing.')
```
